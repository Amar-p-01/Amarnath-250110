```python
import cv2
import numpy as np
```

```python
image = cv2.imread('cameraman_edge_detection.png', cv2.IMREAD_GRAYSCALE)
```

SOBEL

```python
sobel_x=cv2.Sobel(image,cv2.CV_64F,1,0,ksize=3)
sobel_y=cv2.Sobel(image,cv2.CV_64F,1,0,ksize=3)
abs_sobel_x=cv2.convertScaleAbs(sobel_x)
abs_sobel_y=cv2.convertScaleAbs(sobel_y)
sobel_combined=cv2.addWeighted(abs_sobel_x,0.5,abs_sobel_y,0.5,0)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(abs_sobel_x, cmap='gray')
plt.title("Sobel X — Vertical Edges")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(abs_sobel_y, cmap='gray')
plt.title("Sobel Y — Horizontal Edges")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(sobel_combined, cmap='gray')
plt.title("Sobel Combined")
plt.axis('off')

plt.show()
```

**CANNY**

```python
blurred_image = cv2.GaussianBlur(image, (5, 5), 0)
edges=cv2.Canny(blurred_image,threshold1=100,threshold2=200)
```

```python
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.title("Original Grayscale")
plt.imshow(image, cmap='gray')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title("Canny Edge Detection")
plt.imshow(edges, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```

**SCHARR**

```python

scharr_x = cv2.Scharr(image, cv2.CV_64F, 1, 0)
scharr_y = cv2.Scharr(image, cv2.CV_64F, 0, 1)
magnitude = np.sqrt(scharr_x**2 + scharr_y**2)
```

```python
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(magnitude, cmap='gray')
plt.title('Scharr Edge Detection')
plt.axis('off')

plt.tight_layout()
plt.show()
```
