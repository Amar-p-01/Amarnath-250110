Sobel Edge detection

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread('image.jpg')
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY) #type:ignore
plt.imshow(gray_image, cmap='gray')
```

```python
Gx = cv2.Sobel(gray_image, cv2.CV_64F, 1, 0, ksize=5) #type:ignore
Gy = cv2.Sobel(gray_image, cv2.CV_64F, 0, 1, ksize=5) #type:ignore

G = np.sqrt(Gx**2 + Gy**2)

G = np.uint8(255 * G / np.max(G))
```

```python
plt.subplot(2, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB)) #type:ignore
plt.title('Original Image')
plt.axis('off')

# Gradient in X direction
plt.subplot(2, 2, 2)
plt.imshow(Gx, cmap='gray')
plt.title('Gradient in X direction')
plt.axis('off')

# Gradient in Y direction
plt.subplot(2, 2, 3)
plt.imshow(Gy, cmap='gray')
plt.title('Gradient in Y direction')
plt.axis('off')

# Edge-detected image
plt.subplot(2, 2, 4)
plt.imshow(G, cmap='gray')
plt.title('Sobel Edge Detection')
plt.axis('off')

plt.show()
```

With explicit gaussian blurring 

```python
blurred_image = cv2.GaussianBlur(gray_image, (5, 5), 0) #type:ignore

Gx = cv2.Sobel(blurred_image, cv2.CV_64F, 1, 0, ksize=5) #type:ignore
Gy = cv2.Sobel(blurred_image, cv2.CV_64F, 0, 1, ksize=5) #type:ignore

G = np.sqrt(Gx**2 + Gy**2)

G = np.uint8(255 * G / np.max(G))

plt.subplot(2, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB)) #type:ignore
plt.title(f'Original {image.shape[0]}x{image.shape[1]}') #type:ignore
plt.axis('off')

# Gradient in X direction
plt.subplot(2, 2, 2)
plt.imshow(Gx, cmap='gray')
plt.title(f'X direction {Gx.shape[0]}x{Gx.shape[1]}') #type:ignore
plt.axis('off')

# Gradient in Y direction
plt.subplot(2, 2, 3)
plt.imshow(Gy, cmap='gray')
plt.title(f'Y direction {Gy.shape[0]}x{Gy.shape[1]}')
plt.axis('off')

# Edge-detected image
plt.subplot(2, 2, 4)
plt.imshow(G, cmap='gray')
plt.title(f'Sobel {G.shape[0]}x{G.shape[1]}') #type:ignore
plt.axis('off')

plt.show()
```

cv2.Sobel pads internally to preserve the original resolution of the image

Canny Edge detection

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread('image.jpg', cv2.IMREAD_COLOR)
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

blurred_image = cv2.GaussianBlur(gray_image, (5, 5), 1.4)

edges = cv2.Canny(blurred_image, 100, 200)

plt.figure(figsize=(10, 5))

# Original image
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('Original Image')
plt.axis('off')

# Edge-detected image
plt.subplot(1, 2, 2)
plt.imshow(edges, cmap='gray')
plt.title('Canny Edge Detection')
plt.axis('off')

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread('image.jpg', cv2.IMREAD_COLOR)
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

edges = cv2.Canny(gray_image, 60, 300)

plt.figure(figsize=(10, 5))

# Original image
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('Original Image')
plt.axis('off')

# Edge-detected image
plt.subplot(1, 2, 2)
plt.imshow(edges, cmap='gray')
plt.title('Canny Edge Detection')
plt.axis('off')

plt.show()
```
