Using sobel for edge detection

```python
#IMPORTS
import cv2
import matplotlib.pyplot as plt
import numpy as np

#load image
img=cv2.imread("/content/image_edge_detection_sample_application.jpg")

#convert to grayscale
gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)

#defining sobel kernels
sobel_x=np.array([[-1,0,1],[-2,0,2],[-1,0,1]],dtype=np.float32)
sobel_y=np.array([[-1,-2,-1],[0,0,0],[1,2,1]],dtype=np.float32)

#apply kernels
gx=cv2.filter2D(gray,cv2.CV_32F,sobel_x)
gy=cv2.filter2D(gray,cv2.CV_32F,sobel_y)

#take absolute value
gx_abs=np.absolute(gx)
gy_abs=np.absolute(gy)

#gradient magnitude
magnitude=np.sqrt(gx_abs**2+gy_abs**2)

#normalise
magnitude=cv2.normalize(magnitude,None,0,255,cv2.NORM_MINMAX)
magnitude=magnitude.astype(np.uint8)


#results
plt.figure(figsize=(15, 8))

plt.subplot(2, 2, 1)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(gx_abs, cmap="gray")
plt.title("Sobel X")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(gy_abs, cmap="gray")
plt.title("Sobel Y")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(magnitude, cmap="gray")
plt.title("Gradient Magnitude")
plt.axis("off")

plt.show()
```

```python
#IMPORTS
import cv2
import matplotlib.pyplot as plt
import numpy as np

#load image
img=cv2.imread("/content/image_edge_detection_sample_application.jpg")

#convert to grayscale
gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)

#defining sobel kernels
feldman_x=np.array([[-1,0,1],[-1,0,1],[-1,0,1]],dtype=np.float32)
feldman_y=np.array([[-1,-1,-1],[0,0,0],[1,1,1]],dtype=np.float32)

#apply kernels
gx=cv2.filter2D(gray,cv2.CV_32F,feldman_x)
gy=cv2.filter2D(gray,cv2.CV_32F,feldman_y)

#take absolute value
gx_abs=np.absolute(gx)
gy_abs=np.absolute(gy)

#gradient magnitude
magnitude=np.sqrt(gx_abs**2+gy_abs**2)

#normalise
magnitude=cv2.normalize(magnitude,None,0,255,cv2.NORM_MINMAX)
magnitude=magnitude.astype(np.uint8)


#results
plt.figure(figsize=(15, 8))

plt.subplot(2, 2, 1)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(gx_abs, cmap="gray")
plt.title("feldman X")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(gy_abs, cmap="gray")
plt.title("feldman Y")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(magnitude, cmap="gray")
plt.title("Gradient Magnitude")
plt.axis("off")

plt.show()
```

# Experiments

I tried three different edge detection kernels: **Feldman, Sobel, and Scharr**. For each method, I converted the image to grayscale and calculated the horizontal and vertical gradients separately. I then combined them using the gradient magnitude:

### Feldman

Feldman was the simplest of the three. It was able to detect the major edges in the image, but the result was relatively basic and some smaller details were less visible.

### Sobel

Sobel gave a cleaner edge map compared to Feldman. Its extra weighting in the kernel seemed to make the edges more stable while also reducing some small variations.

### Scharr

Scharr produced stronger responses around edges and picked up more fine details in the image. The output was also somewhat more sensitive to small intensity changes.

### Comparison

Overall, all three methods were able to detect the main boundaries in the image, but their outputs were slightly different because of the different kernel weights. Feldman was the simplest, Sobel gave a good balance, and Scharr produced stronger and more detailed responses.

I also experimented with applying Gaussian blur before edge detection. Blurring reduced some unwanted small responses, but too much blur also removed fine edges.

### Conclusion

This experiment helped me understand that edge detection is essentially based on finding changes in pixel intensity. The kernels calculate these changes in the horizontal and vertical directions, and combining the two gradients gives the final edge map.
